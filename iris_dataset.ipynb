```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.datasets import make_classification
```

```python
iris = datasets.load_iris()
```

```python
iris
```

The iris data set contains 4 input features and 1 output variable (the class label)

```python
print(iris.feature_names)
```

output features


```python
print(iris.target_names)
```

```python
print(iris.target)
```

Input features

```python
iris.data
```

```python
iris.target
```

Assigning input and Output variables


```python
x = iris.data #features
y = iris.target #class
```

examine the data dimension

```python
x.shape #(row, col)
```

```python
y.shape
```

Build Classification Model using Random Forest

```python
clf = RandomForestClassifier()
```

```python
clf.fit(x,y) #feature, class
```

Feature Importance

```python
print(clf.feature_importances_)
```

Make Prediction



```python
x[0]
```

```python
print(clf.predict([x[0]]))
```

```python
print(clf.predict_proba(x[[0]]))
```

```python
clf.fit(iris.data, iris.target_names[iris.target])
```

Data split(80/20 ratio)

```python
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size=0.2)
```

```python
x_train.shape, y_train.shape
```

```python
x_test.shape, y_test.shape
```

Rebuild the Random Forest Model

```python
clf.fit(x_train, y_train)
```

performs prediction on single sample from the data set

```python
print(clf.predict([x[0]]))
```

```python
print(clf.predict_proba([x[0]])) # we are 100% sure that it is from the first class
```

performs prediction on the test set

```python
print(clf.predict(x_test)) #predicted class labels
```

```python
print(y_test) #Actual class labels
```

Model performance

```python
print(clf.score(x_test, y_test))
```

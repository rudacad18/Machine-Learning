**linechart**

The line chart is one of the simplest and most widely used data visualization techniques. A line chart displays information as a series of data points or markers connected by straight lines. You can customize the shape, size, color, and other aesthetic elements of the lines and markers for better visual clarity.

Here's a Python list showing the yield of apples (tons per hectare) over six years in an imaginary country called Kanto.

```python
yield_apples = pd.array([0.895, 0.91, 0.919, 0.926, 0.929, 0.931])
```

```python
plt.plot(yield_apples)
```

```python
# plotting two variables
years = [2010, 2011, 2012, 2013, 2014, 2015]
yield_apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931]
```

```python
plt.plot(years, yield_apples)
```

Axis Labels

We can add labels to the axes to show what each axis represents using plt.xlabel and plt.ylabel methods

```python
plt.plot(years, yield_apples)
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
```

Plotting Multiple Lines

You can invoke the plt.plot function once for each line to plot multiple lines in the same graph. Let's compare the yields of apples vs. oranges in Kanto

```python
years = range(2000, 2012)
apples = [0.895, 0.91, 0.919, 0.926, 0.929, 0.931, 0.934, 0.936, 0.937, 0.9375, 0.9372, 0.939]
oranges = [0.962, 0.941, 0.930, 0.923, 0.918, 0.908, 0.907, 0.904, 0.901, 0.898, 0.9, 0.896, ]
```

```python
plt.plot(years, apples)
plt.plot(years, oranges)
plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
```

Chart title and Legend

```python
plt.plot(years, apples)
plt.plot(years, oranges)

plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto')
plt.legend(['Apples', 'Oranges'])
```

Line Markers

We can also show markers for the data points on each line using marker argument in plt.plot .

```python
plt.plot(years, apples, marker = 'o')
plt.plot(years, oranges, marker = 'x')

plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto')
plt.legend(['Apples', 'Oranges'])
```

The fmt argument provides a shorthand for specifying the marker shape, line style, and line color. It can be provided as the third argument to plt.plot.

fmt = '[marker][line][color]'

```python
plt.plot(years, apples, 's-b')
plt.plot(years, oranges, 'o--r')

plt.xlabel('Years')
plt.ylabel('Yield (tons per hectare)')
plt.title('Crop Yields in Kanto')
plt.legend(['Apples', 'Oranges'])
```

```python
plt.plot(years, apples, 'or')
plt.title("Yield of Oranges(tons per hectare)")
```

**Data Visualisation with Seaborn**

Seaborn is a Python data visualization library built on top of the popular Matplotlib library. It provides a high-level interface for creating beautiful and informative statistical graphics, making it an important tool for data analysts, scientists, and machine learning practitioners.

It provides a high-level interface for creating informative and attractive statistical graphics. Seaborn is designed to work with structured data, such as data in Pandas dataframes, and provides functions for visualizing relationships between variables, visualizing distributions of data, and visualizing categorical data.

Some key features of Seaborn include:

High-level interface: Seaborn provides a simple and concise API for creating a ariety of statistical graphics.

Beautiful default styles: Seaborn provides a set of beautiful default styles for creating visually appealing plots.

Flexible plotting functions: Seaborn provides a wide range of functions for visualizing data, including functions for creating scatter plots, line plots, bar plots, and heatmaps.

Statistical analysis tools: Seaborn provides functions for conducting statistical analysis, such as hypothesis testing and regression analysis.

Integration with Pandas: Seaborn integrates well with Pandas, a popular data analysis library in Python, allowing for easy manipulation and visualization of data.

```python
# Load data into a Pandas dataframe
df = sns.load_dataset("iris")
df
```

Scatter Plot

```python
sns.scatterplot(x = 'sepal_length', y = 'sepal_width', data = df)
```

```python
species = df.sepal_width[0] # how to access a parameter
print(species)
```

Histogram

A histogram is a type of plot that displays the distribution of a dataset by dividing it into a set of bins and counting the number of observations that fall into each bin. In Seaborn, a Python data visualization library based on Matplotlib, you can create a histogram using the sns.histplot() function.

```python
sns.histplot(data = df, x = 'petal_width')
plt.xlabel("Petal Width")
plt.ylabel("Frequency")

plt.show()

# we dont need to struggle with y axis, as the y axis in a histogram in itself means frequency w.r.t the parameter in x axis
```

# **Box Plot**

A box plot is a graphical representation of a dataset that summarizes the distribution of a continuous variable.The resulting box plot will display a box that represents the interquartile range (IQR) of the data (the range between the 25th and 75th percentiles), with a line representing the median of the data. The whiskers extend to the minimum and maximum values within 1.5 times the IQR, and any outliers beyond this range are plotted as individual points. The boxplot is a useful tool for visualizing the distribution of data and identifying potential outliers.

To create a box plot in Seaborn, you can use the boxplot() function.

```python
sns.boxplot(df['sepal_length'])
```

```python
plt.figure(figsize= (16,9))

sns.boxplot(x= "petal_length", y = "petal_width", data = df)

plt.xlabel("Length of the petal")
plt.ylabel("Width of the petal")
plt.title("BOX PLOT")
```

Pair Plot
One of the most commonly used visualization functions in Seaborn is the pairplot() function, which creates a matrix of scatterplots and histograms to visualize pairwise relationships in a dataset.

The pairplot() function takes a pandas DataFrame as its input and plots pairwise relationships between all pairs of variables in the dataset. By default, the diagonal shows a histogram of each variable, while the off-diagonal plots show scatterplots of the pairs of variables.

The pairplot() function can also be customized using a variety of parameters to change the appearance and behavior of the plots. For example, you can specify the color of the points, the style of the markers, and the size of the figure. You can also use the hue parameter to color the points based on a categorical variable in the dataset, which can help to reveal patterns and relationships that might not be visible in a simple scatterplot.


```python
# create a pairplot of the iris dataset
sns.pairplot(df)
```

# Heat Map

A heatmap is a graphical representation of data in which values are represented as colors. It is a two-dimensional representation of data that uses color to represent the magnitude of a variable at each point in the plot. Heatmaps are useful for visualizing large datasets and identifying patterns and trends in the data.

In Seaborn, the heatmap function is used to create a heatmap plot. The function takes a rectangular dataset as input and generates a grid of colored cells. The color of each cell represents the value of the corresponding variable in the dataset.

To create a box plot in Seaborn, you can use the heatmap() function.


```python
plt.figure(figsize = (16,9))

#create a pivot table
pivot_table = pd.pivot_table(df, values= 'petal_length', index = ['sepal_length'], columns = ['sepal_width'])

sns.heatmap(pivot_table, cmap= "YlGnBu")
```

```python
newdf = df.drop("species", axis = 'columns')
plt.figure(figsize=(16,9))
sns.heatmap(newdf, cmap= "YlGnBu")
```

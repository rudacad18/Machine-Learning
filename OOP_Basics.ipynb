```python
# constructor

class GameCharacter():

  # Method for setting/ initializing attribute values
  def set_attributes(self, name, weight):
    self.name = name
    self.weight = weight

# creation of two objects
vesimir = GameCharacter()
geralt = GameCharacter()

# setting attributes for vesimir and geralt
vesimir.set_attributes(name = 'Vesimir', weight = 100)
geralt.set_attributes(name = 'Geralt', weight = 150)

print(f"Geralt's weight is {geralt.weight}")
print('\n')
print(f"Vesimir's weight is {vesimir.weight}")
```

Method set_attributes()

It enables setting attribute values for objects.
Unfortunately, we have to call the method every time we create a new object.

Luckily, in Python there is a special method on this case; which is called constructor.

Constructor is a method called automatically when an object is being created.
Besides, it's a special method and has the following syntax:
  __init__(self, params)

In Python methods with underscores belong to a special group of methods called overloading or magic methods.

Inheritance, Polymorphism and Encapsulation

```python
import inspect
# parental class
class GameCharacter:

  #constructor
  def __init__(self, name, hair_color, height):
    self.name = name
    self.hair_color = hair_color
    self.height = height

  #main methods for greeting and goodbye
  def greeting(self):
    print(f"Glad to see you today. My name is {self.name}.")

  def farewell(self):
    print(f"Goodbye.")

# defining a child class

class witcher(GameCharacter): # bracket attribute is compulsory

  # No need for a constructor because it is being inherited from the parent class
  # Define a unique method. Only the witchers have sign abilities.

  def signs_ability(self):
    signs = ['Axii', 'Qven', 'Ignii', 'Aard', 'Yrden']
    return signs

  # create a new object
gerald = witcher(name = 'Geralt', hair_color = 'Grey', height = 200)
print(gerald.signs_ability()) # take note of syntax

# Accessing other attributes from the parental class
print(gerald.name)
print(gerald.greeting())
```

```python
print(issubclass(witcher, GameCharacter)) # to check if child class belongs to a certain parent class
```

Constructor Extension

```python
# we can create __init__() for child class

class witcher(GameCharacter):
  def __init__(self, witcher_school = 'BPHC'):
    self.witcher_school = witcher_school

geralt = witcher()
geralt.witcher_school
```
